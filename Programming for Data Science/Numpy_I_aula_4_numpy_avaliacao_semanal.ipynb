### Exercício 1)
Considere o array `X` como descrito abaixo:
```python
X = np.arange(100).reshape(10,10)
```
Utilizando indexação, gere uma nova matriz `Y`, que contenha apenas as linhas com índices 1,5 e 7 e colunas de 1 a 3 de `X`.
O resultado do comando
```python
print(Y.ravel())
```
será:

a) [11 12 51 52 71 72]<br>
b) [11 12 13 51 52 53 71 72 73]<br>
c) [10 11 12 13 50 51 52 53 70 71 72 73]<br>
d) [11 15 17 21 25 27 31 35 37]

````python
X = np.arange(100).reshape(10,10)
print(X)
print('---')
#Solucao
Y = X[[1, 5, 7,], 1:4]
print(Y.ravel())

# RESPOSTA B
````

### Exercício 3)
Considere o array `A` abaixo:
```python
[[ 2 19 16  1 12  7]
 [ 5 19 18 17  9  3]
 [16  4 18 11  4  2]
 [ 6  2 18 14 15 15]
 [ 4  8  1 14 19 13]]
```
Qual das alternativas abaixo corresponde ao número de dimensões (ndim), a forma (shape) e ao número de elementos (size) no array, respectivamente?

a) (5, 6) 2 2<br>
b) 2 (5, 6) 2<br>
c) (5, 6) 2 30<br>
d) 2 (5, 6) 30

````python
A = np.array([[ 2, 19, 16,  1, 12,  7],
 [ 5, 19, 18, 17,  9,  3],
 [16,  4, 18, 11,  4,  2],
 [ 6,  2, 18, 14, 15, 15],
 [ 4,  8,  1, 14, 19, 13]])

# Solucao
print(A.ndim) #dimensões
print(A.shape) #forma
print(A.size) #elementos

# RESPOSTA D
````

### Exercício 4)
Considere o numpy array `A` do exercício 3). Escreva um código para criar uma máscara booleana para selecionar **todos** os **elementos pares** de `A` e armazená-los em um novo array *unidimensional* `B`. O resultado do comando
```python
print(B.size)
```
será:<br>
a) 13<br>
b) 15<br>
c) 16<br>
d) 17

````python
# Solucao
mask = (A % 2 == 0)
B = A[mask]
print(B.size)

# RESPOSTA C
````

### Exercício 5)
Ainda considerando o array `A` do exercício 3), qual alternativa melhor explica o que o código abaixo está realizando:
```python
l_mask = [(i%2!=0) and (j%2!=0) for i in range(A.shape[0]) for j in range(A.shape[1])]
mask = np.array(l_mask).reshape(A.shape)
A[mask] = -1
```

a) insere elementos -1 nas linhas e colunas pares de `A` utilizando uma máscara booleana<br>
b) insere elementos -1 nas posições com índices ímpares (linha e coluna) de `A` utilizando uma máscara booleana<br>
c) insere elementos -1 nas linhas com índices ímpares e colunas pares de `A` utilizando uma máscara booleana<br>
d) troca os elementos impares da matriz por -1 utilizando uma máscara booleana

````python
l_mask = [(i%2!=0) and (j%2!=0) for i in range(A.shape[0]) for j in range(A.shape[1])]
mask = np.array(l_mask).reshape(A.shape)
A[mask] = -1
print(A)

# RESPOSTA B
````
