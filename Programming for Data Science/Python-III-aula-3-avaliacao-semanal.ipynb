### Exercício 1)
Considere a lista `numeros` declarada na célula abaixo. Utilize o comando <font color='blue'>map</font> para gerar uma nova lista `intervalos` contendo as strings `"x<15"`, `"15<=x<30"' e '"x>=30"' se o elemento da lista `numeros` está no intervalo correspondente. Por exemplo:
```python
numeros = [7, 33, 2, 19,...]
par_impar = ["x<15", "x>=30", "x<15", "15<=x<30",...]
```

Quantas vezes cada string aparece na lista:<br>
a) "x<15": 20, "15<=x<30": 15, "x>=30": 16<br>
b) "x<15": 23, "15<=x<30": 16, "x>=30": 17<br>
c) "x<15": 21, "15<=x<30": 14, "x>=30": 15<br>
d) "x<15": 22, "15<=x<30": 17, "x>=30": 16

````python
numeros = [7, 33, 2, 19, 44, 1, 30, 47, 28, 10, 4, 17, 7, 37, 21, 32, 44, 2, 33, 9, 26,
           9, 29, 9, 49, 11, 8, 42, 26, 23, 17, 16, 37, 26, 19, 26, 8, 27, 15, 10, 31,
           41, 42, 10, 4, 9, 7, 18, 44, 12, 33, 13, 12, 5, 6, 36]
````

````python
# Solucao
def categorizar(x):
    if x < 15:
        return "x<15"
    elif 15 <= x < 30:
        return "15<=x<30"
    else:
        return "x>=30"

# Aplicando o map para gerar a nova lista
intervalos = list(map(categorizar, numeros))

# Contagem das ocorrências
contagem = {
    "x<15": intervalos.count("x<15"),
    "15<=x<30": intervalos.count("15<=x<30"),
    "x>=30": intervalos.count("x>=30")
}

print(contagem)
# RESPOSTA B
````

### Exercício 2)
Considere a lista `numeros` do exercício 1). Utilize o comando <font color='blue'>filter</font> para gerar uma nova lista contendo somente os elementos da lista `numeros` que são maiores ou iguais a 30 ou estão entre 10 e 20 (estritamente maiores que 10 e menores que 20). A soma dos valores na lista resultante é:<br>
a) 824<br>
b) 140<br>
c) 651<br>
d) 737<br>

````python
# Solucao
lista_filtrada = list(filter(lambda x: x >= 30 or (10 < x < 20), numeros))

soma = sum(lista_filtrada)
print(soma)
# RESPOSTA A
````

### Exercício 3)
Assinale a alternativa que melhor explica o que significa `f` e `x` na declaração `f(x)` que está dentro do comando <font color='blue'>map</font> no código abaixo:<br>
a) `x` corresponde a lista de números [10, 20, 30] e `f` é um dos elementos desta lista.<br>
b) `f` corresponde a lista `funcoes` e `x` é um dos elementos desta lista.<br>
c)  `f` corresponde a um números da lista [10, 20, 30] e `x` é um dos elementos da lista `funcoes`.<br>
d) `f` corresponde a um dos elementos da lista `funcoes` e `x` é um número da lista [10, 20, 30].

````python
def subtracao(x):
    return(x-1)

def soma(x):
    return(x+x)

def multiplica(x):
    return(x*x)

funcoes = [subtracao, soma, multiplica]
numeros = [10, 20, 30]

for x in numeros:
    res = list(map(lambda f: f(x),funcoes))
    print(res)

# RESPOSTA D
````

### Exercício 5)
A classe <font color='blue'>`EquacaoLinear`</font> definida na célula abaixo representa uma equação linear de primeiro grau na forma <font color='blue'>`f(x) = a*x + b`</font>. Os códigos descritos nas opções `(I)`, `(II)` e `(III)` a abaixo apresentam três formas distintas de instanciar os coeficiêntes `a` e `b` da equação com valores `a=1` e `b=2`, respectivamente.<br>
<br>
(I)
```python
el = EquacaoLinear(a=1, b=2)
```
II)
```python
el = EquacaoLinear()
el.atualiza_coeficientes(a=1, b=2)
```
III)
```python
el = EquacaoLinear(a=1)
el._b = 2
```

Assinale a alternativa correta:<br>
a) Apenas `(I)` instancia corretamente as variáveis.<br>

b) Apenas `(II)` e `(III)` instanciam corretamente as variáveis.<br>

c) Apenas `(I)` e `(III)` instanciam corretamente as variáveis.<br>

d) `(I)`, `(II)` e `(III)` instanciam corretamente as variáveis.<br>

````python
class EquacaoLinear(object):

  def __init__(self, a=None, b=None):
    self._a = a
    self._b = b

  def atualiza_coeficientes(self, a, b):
    self._a = a
    self._b = b

  def resolve(self, x):
    return(self._a*x + self._b)

# RESPOSTA D
````
