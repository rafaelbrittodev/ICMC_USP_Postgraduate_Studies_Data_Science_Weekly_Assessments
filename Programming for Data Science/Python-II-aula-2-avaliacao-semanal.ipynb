### Exercício 2)

a) graduado<br>

````python
# Solucao
dc = {}

for cargo, formacao, nome in mat:
    if formacao not in dc:
        dc[formacao] = []

    if cargo not in dc[formacao]:
        dc[formacao].append(cargo)

print("Dicionário dc:", dc)

for formacao, cargos in dc.items():
    print(f"{formacao}: {len(cargos)} cargos -> {cargos}")

# Resposta B
````

### Exercício 3)
A função `variacoes` descrita abaixo
```python
def variacoes(s,c):
    '''s é uma string
       c é um caractere
    '''
    l = {______________}
    return(l)
```
recebe uma 'string' e um caractere como parâmetros, gerando uma lista como resposta, onde os elementos da lista são todas as variações da string com uma ocorrência do caractere enviado como parâmetro removido em cada variação. Por exemplo:
```python
print(variacoes('oscasados','s'))
```
deve resultar em:
```python
['ocasados', 'oscaados', 'oscasado']
```

Qual o tamanho da lista resultante quando a palavra "pneumoultramicroscopicossolicovulconoconiotico" e o caractere "o" são passados como parâmetros para a função `variacoes`?

a) 8<br>
b) 9<br>
c) 10<br>
d) 11

````python

def variacoes(s, c):
    '''s é uma string
       c é um caractere
    '''
    # Para cada índice 'i' em que s[i] for igual a 'c',
    # concatenamos a parte antes de 'i' com a parte depois de 'i'.
    l = [s[:i] + s[i+1:] for i in range(len(s)) if s[i] == c]
    return l

# Teste com a palavra do exercício
palavra = "pneumoultramicroscopicossolicovulconoconiotico"
caractere = "o"

res = variacoes(palavra, caractere)
print(f"Lista resultante: {res}")
print(f"Tamanho da lista: {len(res)}")

# Resposta D
````

O método <font color='blue'>sorted</font> assume como atributos uma sequência e um parâmetro `key` que pode ser uma função que será aplicada aos elementos da sequência a fim de ordená-los de acordo com os valores resultantes da função (veja lista de exercícios da aula).

Utilize uma função <font color='blue'>lambda</font> como parâmetro `key` no método <font color='blue'>sorted</font> para ordenar uma lista cujos elementos são numéricos, de modo que cada elemento seja ordenado como se os valores correspondessem a $x^2$, onde $x$ corresponde a cada elemento da lista.

Por exemplo, a lista
```python
[-0.11, -0.57, 0.85, -0.78, 0.13]
```
seria ordenada como:
```python
[-0.11, 0.13, -0.57, -0.78, 0.85]
```

Qual o primeiro e o último elemento da lista ordenada quando a ordenação é aplicada à:
```python
[0.10, -0.11, 0.4, 0.12, -0.57, -0.05, 0.85, -0.27, -0.07, -0.78, 0.13, 0.36]
```

a) 0.4 e -0.78<br>
b) 0.36 e -0.78<br>

````python
l = [0.10, -0.11, 0.4, 0.12, -0.57, -0.05, 0.85, -0.27, -0.07, -0.78, 0.13, 0.36]
#l = [-0.11, -0.57, 0.85, -0.78, 0.13]

# Solucao
# Ordenando com a chave lambda x**2
l_ordenada = sorted(l, key=lambda x: x**2)

print("Lista ordenada:", l_ordenada)
print(f"Primeiro elemento: {l_ordenada[0]}")
print(f"Último elemento: {l_ordenada[-1]}")

# Resposta C
````

### Exercício 5)
Construa uma função `concatena_dicionarios` que recebe dois dicionários como atributos, onde as chaves e valores de ambos são números inteiros. A função deve concatenar ambos os dicionários em um único dicionário contendo as chaves e os valores de ambos os dicionários. Se a chave se repete em ambos, mantenha o maior valor associado a chave. Por exemplo, dados os dicionários:
```python
dc1 = {1: 5479, 5: 1479, 7: 1549, 9: 1547}
dc2 = {5: 2647, 2: 9647, 1: 9247}
```
a função deve resultar no dicionário
```python
{1: 9247, 7: 1549, 5: 2647, 9: 1547, 2: 9647}
```
Considerando os dicionários abaixo como entrada para a função:
```python
dc1 = {4: 10153, 18: 41118, 2: 101515, 15: 354, 7: 4145, 8: 10316, 3: 121016, 11: 15122, 12: 16158, 10: 4188, 14: 15318, 5: 11711, 16: 181412}

dc2 = {19: 16619, 4: 11310, 18: 171112, 11: 171817, 12: 171911, 6: 4195, 5: 171910, 3: 10310, 16: 191916, 10: 111619, 17: 51016}
```
o par `chave:valor` com o maior valor no dicionário concatenado será:

a) 12: 171911<br>
b) 5: 171910<br>
c) 11: 171817<br>
d) 16: 191916

````python
dc1 = {4: 10153, 18: 41118, 2: 101515, 15: 354, 7: 4145, 8: 10316,
       3: 121016, 11: 15122, 12: 16158, 10: 4188, 14: 15318, 5: 11711,
       16: 181412}

dc2 = {19: 16619, 4: 11310, 18: 171112, 11: 171817, 12: 171911,
       6: 4195, 5: 171910, 3: 10310, 16: 191916, 10: 111619, 17: 51016}
````

````python
def concatena_dicionarios(d1,d2):
    # Solucao
    # Cópia do primeiro dicionário
    resultado = d1.copy()

    for chave, valor in d2.items():
        if chave in resultado:
            # Se a chave existe, mantemos o maior valor
            resultado[chave] = max(resultado[chave], valor)
        else:
            # Se não existe, apenas adicionamos
            resultado[chave] = valor

    return resultado

# Executando a função
dc_final = concatena_dicionarios(dc1, dc2)

# Encontrando o par chave:valor com o maior valor total
maior_chave = max(dc_final, key=dc_final.get)
maior_valor = dc_final[maior_chave]

print(f"Dicionário concatenado: {dc_final}")
print(f"O par com o maior valor é {maior_chave}: {maior_valor}")

# Resposta D
````
