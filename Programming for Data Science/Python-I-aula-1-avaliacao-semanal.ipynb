### Exercício 1
Considere a lista `lt` abaixo. Escreva um código para calcular quantos elementos da lista são divisíveis por 3.

a) Número de elementos divisiveis por 3 = 15<br>
b) Número de elementos divisiveis por 3 = 13<br>
c) Número de elementos divisiveis por 3 = 10<br>

````python
lt = [99, 128, 185, 19, 215, 493, 104, 78, 248, 299, 278, 141, 467, 212, 465, 130, 320, 35, 148, 351,
       176, 366, 373, 256, 314, 354, 188, 259, 443, 81, 447, 313, 368, 197, 185, 328, 275, 269, 153,
       495, 155, 312, 87, 141, 413, 129, 106, 44, 245, 479]

# escreva o código para encontrar a solução
soma_div3 = 0


for i in lt:
  if i % 3 == 0:
    soma_div3 += 1

print(soma_div3)
#Resposta correta A
````

### Exercício 2

### Exercício 4
Escreva um código para encontrar qual estado possui o maior valor total de reembolsos, ou seja, somando os reembolsos de todos os deputados de cada estado, qual estado possui o maior valor de reembolso? Qual é o estado e o valor correspondente?

__Dica:__ Crie um dicionário onde as chaves são os estados e os valores são o total de reembolsos do estado correspondente.

a) SP, 9546.60<br>
b) MA, 19456.87<br>
c) MT, 11132.70<br>
d) PE, 12056.00

````python
# escreva o código para encontrar a solução
somas_por_estado = {}

for i in dados_deputados:
    estado = i[1]
    gasto = i[3]

    if estado in somas_por_estado:
        somas_por_estado[estado] += gasto
    else:
        somas_por_estado[estado] = gasto

somas_ordenadas = sorted(somas_por_estado.items(), key=lambda item: item[1], reverse=True)

for estado, total in somas_ordenadas:
    print(f"{estado}: R$ {total:.2f}")
# Resposta correta D
````

### Exercício 5
Escreva um código para encontrar qual partido possui o maior número de deputados na lista apresentada.
Qual o partido e a quantidade de deputados vinculados a ele?

a) PTB, 3 vezes<br>
b) PT, 2 vezes<br>
c) PSDB, 6 vezes<br>
d) PMDB, 5 vezes

````python
# escreva o código para encontrar a solução
soma_por_partido = {}

for i in dados_deputados:
    partido = i[2]

    if partido in soma_por_partido:
        soma_por_partido[partido] += 1
    else:
        soma_por_partido[partido] = 1

somas_ordenadas = sorted(soma_por_partido.items(), key=lambda item: item[1], reverse=True)

for partido, total in somas_ordenadas:
    print(f"{partido}: {total}")
# Resposta correta C
````
